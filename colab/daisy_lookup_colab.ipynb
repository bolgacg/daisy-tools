# DAISY with one Wikipedia lookup

Runs the group's DAISY task and the same task with one lookup, in the dfm-evals (Inspect AI) format, on a free Colab GPU. Runtime: T4. About 25 minutes for the index, then about 40 minutes for the 592 questions.

Not yet tested from Colab: the repository must be public first.
